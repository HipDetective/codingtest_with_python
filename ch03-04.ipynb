## 03-04_정렬 기초 다지기

### 오름차순 정렬

```python
# 원본 리스트 선언
A = [5, 3, 2, 4, 1]

# 방법1: list.sort() 메서드 사용
A.sort()
print("sort() 결과:", A)

# 원본 리스트 초기화
A = [5, 3, 2, 4, 1]

# 방법2: sorted() 정렬된 복사본을 반환(원본 유지)
B = sorted(A)
print(f'원본 리스트 {A}')
print(f'sorted() 결과: {B}')
```

```python
### 내림차순 정렬
```

```python
A = [5, 3, 2, 4, 1]

# 리스트 자체를 내림차순 정렬
A.sort(reverse=True)
print(f'내림차순 결과: {A}')

# 원본 리스트 초기화
A = [5, 3, 2, 4, 1]

# 정렬된 복사본 반환
B = sorted(A, reverse=True)
print(f'내림차순 복사본: {B}')
```

```python
# 내림차순 정렬2 - 부호 전환 테크닉
A = [5, 3, 2, 4, 1]

# 부호 반전 -> 오름차순 정렬 -> 부호 되돌리기
A = [-x for x in A]
A.sort()
A = [-x for x in A]

print(f'부호 반전 방식 내림차순: {A}')
```
