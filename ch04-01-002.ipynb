#### 04-01-002 평균 구하기

```python
# 슈도코드
# n에 과목의 수 입력
# mylist에 점수 정보 저장
# mymax에 mylist 정보 중 최대값 저장
# sum에 mylist 모든 데이터 값 더하기
# sum * 100 / mymax / n 출력
```

```python
# 코드 구현
n = input()
mylist = list(map(int, input().split( )))
print(mylist)
mymax = 0
for _ in mylist:
    mymax = _ if mymax < _ else mymax
print(mymax)
mysum = sum(mylist)
print(f"{mymax=}, {mysum=}")
print( mysum / mymax * 100 / int(n))
```
