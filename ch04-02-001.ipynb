#### 04-02-003_구간 합 구하기1

```python
# 슈도 코드
# suNo(숫자 개수), quizNo(질의 개수)
# numbers 변수에 숫자 데이터 저장
# prefix_sum 합 배열 변수 선언
# temp 변수 선언

# for 저장한 숫자 데이터 차례대로 탐색:
#     temp에 현재 숫자 데이터 더하기
#     합 배열에 temp값 저장

# for 질의 개수만큼 반복:
#     질의 범위 받기(s ~ e)
#     구간 합 출력하기(prefix_sum[e] - prefix_sum[s - 1])
```

```python
# 코드 구현
import sys
# input = sys.stdin.readline
suNo, quizNo = map(int, input('배열숫자와 질의숫자를 입력하세요.').split())
print(f'배열의 길이는 {suNo}이며, 질의숫자는 {quizNo}입니다.')
numbers = list(map(int, input().split()))
prefix_sum = [0]
print(f'입력한 배열: {numbers}')
temp = 0

# 합 배열 만들기
for i in numbers:
    temp += i
    prefix_sum.append(temp)

# 합 배열을 통해 구간합 구하기
for idx, i in enumerate(range(quizNo)):
    print(f'회차: {idx + 1}')
    start, end = map(int, input("배열의 시작과 끝을 입력하세요: ").split())
    print(prefix_sum[end] - prefix_sum[start - 1])
```
