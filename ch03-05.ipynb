## 03-05_다중 정렬 조건 익히기

### 튜플 기반 정렬

```python
# 영어, 수학 순서로 튜플 데이터 저장
scores = [
    (80, 100),
    (100, 50),
    (70, 100),
    (80, 90)
]

# 영어 내림차순, 수학 내림차순 순
scores.sort(key=lambda x: (-x[0], -x[1]))

for s in scores:
    print(f'영어={s[0]}, 수학={s[1]}')
```

###  딕셔너리 기반 정렬

```python
# 딕셔너리 기반 다중 조건 정렬
scores = [
    {'english':80, 'math':100},
    {'english':100, 'math':50},
    {'english':70, 'math':100},
    {'english':80, 'math':90}
]

# 수학 내림차순, 영어 내림차순 순
scores.sort(key=lambda x: (-x['math'], -x['english']))

for s in scores:
    print(s)
```
