### 문제001_숫자의 합 구하기

```python
# 수도코드
## n값 입력받기
## numbers 변수에 list 함수를 이용하여 숫자를 하나씩 나누어 받기
## sum 변수 선언
##
## numbers 탐색
##     sum 변수에 numbers에 있는 각 자릿수를 가져와 더하기
##
## sum 출력
```

```python
# 코드 구현
n = input()
numbers = list(input())
sum = 0

for _ in numbers:
    sum += int(_)

print(sum)
```
