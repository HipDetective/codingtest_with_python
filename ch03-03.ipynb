## 03-03_나머지 연산의 중요성 알아보기

```python
A, B, C = 20, 6, 3
```

```python
# 덧셈의 분배 법칙
assert (20 + 6) % 3 == 26 % 3
assert (20 % 3 + 6 % 3) == (2 + 0) % 3
```

```python
# 뺄셈의 분배 법칙
assert (20 - 6) % 3 == 14 % 3
assert (20 % 3 - 6 % 3) == (2 - 0)
```

```python
# 곱셈의 분배 법칙
assert (20 * 6) % 3 == (120) % 3
assert (20 % 3) * (6 % 3) == (2 * 0)
```

```python
# 나눗셈의 분배 법칙 - 미성립
assert (20 // 6) % 3 == 3 % 3
assert (20 % 3) // (6 % 3) == 2 // 2
```

```python
# 1부터 100000까지 곱한 값을 1000000007로 나눈 나머지를 구하기 - 방법1
import time

divisor = 1000000007
answer = 1
start = time.time()

for i in range(1, 100001):
    answer *= i

result = answer % divisor
end = time.time()

print('결과:', result)
print(f'수행 시간: {end - start:.6f}초')
```

```python
# 1부터 100000까지 곱한 값을 1000000007로 나눈 나머지를 구하기 - 방법2
# 수가 커지면 연산 속도가 느려질 수 있음
import time


MOD = 1000000007
answer = 1
start = time.time()

for i in range(1, 100001):
    answer = (answer * i) % MOD

end = time.time()
print('결과:', answer)
print(f'수행 시간: {end - start:.6f}초')
```
